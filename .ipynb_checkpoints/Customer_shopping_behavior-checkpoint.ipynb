```python
# Checking Null Values
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.rename(columns={
    'ï»¿Customer ID':'Customer ID'
},inplace=True)
```

```python
df.columns
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df.rename(columns={
    'purchase_amount_(usd)':'purchase_amount'
},inplace = True)
```

```python
df.columns
```

```python
# create a column age_groub
labels = ['Young Adult', 'Adult','Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels )
```

```python
df[['age','age_group']].head(10)
```

```python
# create columns purchase_freequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14 ,
    'Annually' : 365 ,
    'Every 3 Months' : 90
}

df['purchase_freequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_freequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# drop the douplicated columns
df = df.drop('promo_code_used',axis=1)
```

```python
import pyodbc
print(pyodbc.drivers())
```

```python
import pandas as pd
from sqlalchemy import create_engine
import urllib

# 1. Connect
params = urllib.parse.quote_plus(
    "DRIVER={ODBC Driver 18 for SQL Server};"
    "SERVER=DESKTOP-8B44M08;"
    "DATABASE=Python;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)
engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={params}",
    fast_executemany=True  # 🔥 លឿន
)

# 2. Upload
df.to_sql("sales_data", engine, if_exists="replace", index=False)
print("Done!")

```

```python
df_sql = pd.read_sql("select * from sales_data", engine)
df_sql.head(10)
```
